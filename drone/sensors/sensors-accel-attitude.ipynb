## Accelerometer and Roll/Pitch

The accelerometer measures specific forces on the body ($f_b$):

$$
\begin{align}
\textbf{f}^{b} &= \begin{bmatrix} f_x \\ f_y \\ f_z \end{bmatrix} = \textbf{C}_{n}^{b} \begin{bmatrix} 0\\0\\-g \end{bmatrix} = \begin{bmatrix} g\sin\theta \\ -g\cos\theta\sin\phi \\ -g\cos\phi\cos\theta \end{bmatrix}
\end{align}
$$

where $C_{n}^{b}$ is the rotation matrix rotating the body frame to the navigation frame. Given this relationship, you can calculate the roll ($\phi$) and pitch($\theta$) of the body by:

$$
\begin{align}
\frac {f_y} {f_z} &= \frac {-g\cos\theta\sin\phi} {-g\cos\phi\cos\theta} \\
     &= \frac {g\sin\theta} {g\cos\theta} \\
     &= \tan \phi \\  
\phi &= atan2(-f_y, -f_z) \\
\\
\frac {f_x} {\sqrt{f_y^2 + f_z^2}} &= \frac {g\sin\theta}{\sqrt{g^2\cos^2\theta(\sin^2\phi + \cos^2\phi)}} \\
                                   &= \frac {g\sin\theta} {g\cos\theta} = \tan \theta \\
                            \theta &= atan2(f_x, \sqrt{f_x^2 + f_z^2})
\end{align}
$$

```python
import numpy as np

def accelLevel(accelRead):
    # Takes accelerometer specific force reading when stationary and estimates
    # phi and theta from the gravity vector
    fx_b = accelRead[0]
    fy_b = accelRead[1]
    fz_b = accelRead[2]
    phi_est = np.arctan2(-fy_b,-fz_b)
    theta_est = np.arctan2(fx_b, np.sqrt(fy_b*fy_b + fz_b*fz_b))
    return phi_est, theta_est
```
